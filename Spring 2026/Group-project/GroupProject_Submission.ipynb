### ⚙️ Setup — run this first

```python
import json, os, time, tempfile
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# ── Must match the instructor notebook ──────────────────────────
SUBMISSIONS_FILE = '/home/jovyan/shared-readwrite/group_project/submissions.json'
# ────────────────────────────────────────────────────────────────

DATASETS = [
    '--- Select a dataset ---',
    'DS1  · Ecological Footprint by Country',
    'DS2  · Heart Disease',
    'DS3  · Periodic Table',
    'DS5  · Air Quality',
    'DS6  · East Coast Earthquakes',
    'DS7  · Pennypack Creek Stream Gauge',
    'DS8  · Philadelphia Weather Data',
    'DS9  · Fetal Health',
    'DS10 · Diabetes Prediction',
    'DS11 · Philly School Graduation Rates',
    'DS12 · Jeopardy Dataset',
    'DS14 · Global Sustainable Energy Production',
    'DS16 · Gene Expression in Cancer vs. Normal Tissue',
    'DS17 · Human Microbiome Diversity',
    'DS18 · Student Performance in Secondary Education',
    'DS20 · Protein Amino Acid Composition',
    'CAMELS · Hydrology Dataset',
]

ROLES = ['Coder 1', 'Coder 2', 'Analyst', 'Context Expert', 'Writer', 'Project Manager', 'Visualizer']

# The shared directory must already exist (created by instructor via Dashboard notebook)
shared_dir = os.path.dirname(SUBMISSIONS_FILE)
if not os.path.isdir(shared_dir):
    print(f'⚠️  Shared folder not found: {shared_dir}')
    print('   Ask your instructor to run the Dashboard notebook first.')
else:
    print('✅ Setup complete — fill in your group details below')
```

---
### 📝 Step 1 — Group Identity

```python
style_lbl = {'description_width': '160px'}
layout_wide  = widgets.Layout(width='600px')
layout_full  = widgets.Layout(width='600px', height='100px')

w_team = widgets.Text(
    description='🏷️ Group Name:',
    placeholder='e.g. Broad Street Bootstrappers',
    style=style_lbl, layout=layout_wide
)
w_title = widgets.Text(
    description='📌 Project Title:',
    placeholder='e.g. Microbiome Diversity in Health vs. Disease',
    style=style_lbl, layout=layout_wide
)
w_dataset = widgets.Dropdown(
    description='📂 Dataset:',
    options=DATASETS,
    style=style_lbl, layout=widgets.Layout(width='600px')
)
w_hypothesis = widgets.Textarea(
    description='🔬 Hypothesis:',
    placeholder='State your testable hypothesis, e.g. "Gut microbiome diversity (Shannon index) is lower in individuals with inflammatory bowel disease compared to healthy controls."',
    style=style_lbl, layout=widgets.Layout(width='600px', height='90px')
)
w_description = widgets.Textarea(
    description='📖 Description:',
    placeholder='Brief description of your project plan: what you will explore, what statistical methods you plan to use, and why this question matters.',
    style=style_lbl, layout=widgets.Layout(width='600px', height='110px')
)

display(
    widgets.HTML('<div style="font-weight:600; color:#9E1B34; margin-bottom:6px;">Group Identity</div>'),
    w_team, w_title, w_dataset,
    widgets.HTML('<div style="font-weight:600; color:#9E1B34; margin: 10px 0 6px;">Research Plan</div>'),
    w_hypothesis, w_description
)
```

---
### 👥 Step 2 — Team Members & Roles
Add each member's name and their primary role.

```python
MAX_MEMBERS = 6
member_widgets = []

for i in range(MAX_MEMBERS):
    name_w = widgets.Text(
        description=f'Member {i+1} Name:',
        placeholder='First Last',
        style=style_lbl,
        layout=widgets.Layout(width='320px')
    )
    role_w = widgets.Dropdown(
        description='Role:',
        options=['---'] + ROLES,
        style={'description_width': '45px'},
        layout=widgets.Layout(width='240px')
    )
    member_widgets.append((name_w, role_w))
    display(widgets.HBox([name_w, role_w], layout=widgets.Layout(gap='12px')))

print('(Leave unused member rows blank)')
```

---
### 🚀 Step 3 — Submit!

```python
btn_submit = widgets.Button(
    description='🚀 Submit Project Proposal',
    layout=widgets.Layout(width='260px', height='44px'),
    style={'button_color': '#9E1B34', 'font_weight': 'bold'}
)
out = widgets.Output()

def on_submit(_):
    with out:
        clear_output(wait=True)
        
        # --- Validate ---
        team = w_team.value.strip()
        title = w_title.value.strip()
        dataset = w_dataset.value
        hypothesis = w_hypothesis.value.strip()
        description = w_description.value.strip()
        
        errors = []
        if not team:
            errors.append('⚠️ Group Name is required')
        if not title:
            errors.append('⚠️ Project Title is required')
        if dataset == DATASETS[0]:
            errors.append('⚠️ Please select a dataset')
        if not hypothesis:
            errors.append('⚠️ Hypothesis is required')
        if not description:
            errors.append('⚠️ Project Description is required')
        
        members = []
        for name_w, role_w in member_widgets:
            n = name_w.value.strip()
            r = role_w.value
            if n:
                members.append({'name': n, 'role': r if r != '---' else 'Unassigned'})
        
        if not members:
            errors.append('⚠️ At least one team member is required')
        
        if errors:
            display(HTML('<div style="color:#9E1B34;font-weight:bold;">' + '<br>'.join(errors) + '</div>'))
            return
        
        # --- Build entry ---
        entry = {
            'team': team,
            'title': title,
            'dataset': dataset,
            'hypothesis': hypothesis,
            'description': description,
            'members': members,
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'submitted_at': time.strftime('%H:%M')
        }
        
        # --- Atomic write ---
        try:
            with open(SUBMISSIONS_FILE) as f:
                subs = json.load(f)
        except (FileNotFoundError, json.JSONDecodeError):
            subs = []
        
        # Replace existing submission by same team
        subs = [s for s in subs if s.get('team', '').lower() != team.lower()]
        subs.append(entry)
        
        dir_ = os.path.dirname(SUBMISSIONS_FILE)
        os.makedirs(dir_, exist_ok=True)
        with tempfile.NamedTemporaryFile('w', dir=dir_, delete=False, suffix='.tmp') as f:
            json.dump(subs, f, indent=2)
            tmp = f.name
        os.replace(tmp, SUBMISSIONS_FILE)
        
        # --- Confirmation card ---
        member_rows = ''.join(
            f'<tr><td style="padding:3px 8px;color:#555;">{m["name"]}</td>'
            f'<td style="padding:3px 8px;color:#9E1B34;font-weight:500;">{m["role"]}</td></tr>'
            for m in members
        )
        display(HTML(f'''
        <div style="background:#F5F2EC;border:2px solid #9E1B34;border-radius:10px;padding:20px 24px;max-width:580px;font-family:system-ui,sans-serif;">
          <div style="font-size:1.4em;">✅ <strong style="color:#9E1B34;">Proposal Submitted!</strong></div>
          <div style="margin:12px 0 6px;font-size:12px;color:#888;">Submitted at {entry["timestamp"]}</div>
          
          <table style="width:100%;border-collapse:collapse;margin-top:10px;">
            <tr><td style="padding:4px 0;font-weight:600;color:#444;width:130px;">Group:</td><td style="color:#1A1226;font-weight:700;font-size:1.05em;">{team}</td></tr>
            <tr><td style="padding:4px 0;font-weight:600;color:#444;">Title:</td><td style="color:#333;font-style:italic;">{title}</td></tr>
            <tr><td style="padding:4px 0;font-weight:600;color:#444;">Dataset:</td><td style="color:#333;">{dataset}</td></tr>
          </table>
          
          <div style="margin-top:12px;font-weight:600;color:#444;font-size:13px;">Team Members:</div>
          <table style="margin-top:4px;">{member_rows}</table>
          
          <div style="margin-top:12px;background:#fff;border-radius:6px;padding:10px 12px;">
            <div style="font-weight:600;color:#9E1B34;font-size:12px;">HYPOTHESIS</div>
            <div style="font-size:13px;color:#333;margin-top:4px;font-style:italic;">{hypothesis}</div>
          </div>
        </div>
        '''))

btn_submit.on_click(on_submit)
display(btn_submit, out)
```
