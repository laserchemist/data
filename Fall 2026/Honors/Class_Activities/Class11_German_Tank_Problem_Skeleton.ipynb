## Part 4. Which Estimator Would You Take Into a War?

Now put numbers on what you just saw. For each estimator you need five things: the **mean**
and the **median** (where the distribution sits), the **standard deviation** and the
**interquartile range** (how wide it is), and the fraction of estimates that landed **within
20% of the truth** — between 800 and 1200.

The first four are single function calls. The last one uses a trick worth knowing:
comparing an array to a number gives you back an array of `True`/`False` values, and the
*average* of those is the fraction that are `True`.

```python
# What a comparison does to an array
small_example = make_array(950, 1300, 1010, 700, 1150)

small_example >= 800
```

```python
# ...and averaging the True/False values gives the fraction that passed
np.mean(small_example >= 800)
```

To ask for two conditions at once, use `&` and put parentheses around each condition.
(The word `and` does not work on arrays — only `&` does.)

```python
# Worked model: the five summaries for est_max
print('mean      ', np.round(np.mean(max_estimates), 1))
print('median    ', np.round(np.median(max_estimates), 1))
print('SD        ', np.round(np.std(max_estimates), 1))
print('IQR       ', np.round(np.percentile(max_estimates, 75)
                             - np.percentile(max_estimates, 25), 1))

close_enough = (max_estimates >= 800) & (max_estimates <= 1200)
print('within 20%', np.round(np.mean(close_enough), 3))
```
