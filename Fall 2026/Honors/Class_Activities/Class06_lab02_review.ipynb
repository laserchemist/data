## Cleaning Data
Many times the files you read do not have the data in a format that is easy for Python to load as numbers. In such cases you need to "clean" the data. **The old joke is that in data science you spend 90% of your time cleaning the data and the othe 10% complaining about cleaning the data.**


Here the goal is to extract Iris's weight in pounds as a number from the string.

```python
baby_wt = "Iris weighed 8lbs 9oz at birth."
baby_wt
```

```python
baby_wt = baby_wt.split()
baby_wt
```

```python
lbs = baby_wt[2]
lbs
```

```python
lbs = lbs.replace("lbs","")
lbs
```

```python
lbs = float(lbs)
lbs
```

## Chaining steps together
All of this can be accomplished in one line of code, but the more you chain the less readable your code becomes Use your judgement.

```python
baby_wt = "Iris weighed 8lbs 9oz at birth."

lbs = float(baby_wt.split()[2].replace("lbs",""))
lbs
```

## Importing modules
When you import a module you are just loading function and constants into Python that you want to use to solve some problem.

```python
# pi is defined in the math module so if we need some pi, we can import it.
import math
math.pi
```

```python
# You can import the module with an alias
import math as m
m.pi
```
