## A smart way to program -- create tests before you write your function.
We are going to create a function that returns the average of a list of numbers.

## A suite of checks
Notice that we have a variety tests. You could easily add more.

```python
def test_average():
    assert average([2, 4, 6]) == 4
    assert average([10, 20, 30, 40]) == 25
    assert average([1, 1, 1, 1, 1]) == 1
    assert average([100]) == 100   # a single-value list 
    print("  all tests passed.")
```

## Function has a bug

```python
def average(values):
    """Return the mean of a list of numbers."""
    for v in values:
        total = 0            # bug: this line belongs before the loop
        total = total + v
    return total / len(values)
```

```python
test_average()
```

## Fix the bug

```python
def average(values):
    """Return the mean of a list of numbers."""
    total = 0
    for v in values:
        total = total + v
    return total / len(values)
```

```python
test_average()
```
