### 1.8 Sampling from a Population

So far in this section we've been **simulating** — using `np.random.choice()` to generate brand-new random data from a model (a die roll plus a modifier). **Sampling** is a related but different idea: instead of generating new data, we draw a smaller *sample* out of an existing, larger collection of data called the **population**. The `datascience` `Table` has a built-in `.sample()` method for exactly this. This is the core idea behind [Chapter 10: Sampling and Empirical Distributions](https://inferentialthinking.com/chapters/10/Sampling_and_Empirical_Distributions.html).

Let's build a large simulated "population" of Roga's door-knocking attempts — think of this as if Roga had attempted the action 10,000 times over a long adventuring career — and then practice drawing samples from it.

````python
# Treat this large simulated collection as our "population"
population_size = 10000
population_rolls = Table().with_columns('modified_roll', simulate_observations(modifier, population_size))
population_rolls
````
