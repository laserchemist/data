## 🧭 Before you write: what goes in a proposal
| Part | What a strong one has |
|---|---|
| 📌 **Title** | Short and specific: a reader knows what you are studying. |
| ❓ **Research question** | One main question your data can actually answer, naming the variables and the groups or places. *"Is the air worse in summer?"* is vague; *"Is the average daily PM2.5 in Philadelphia higher in July than in January?"* can be tested. |
| 📖 **Literature (at least 3)** | Published work your study builds on: peer-reviewed articles, scientific reports, government studies. For each: the full citation, a link or DOI, and **one line on how it supports your study**: what is already known, why your question matters, or which method others used. |
| 🗺️ **Research plan** | How you will research the topic: background you will read, which variables you will use, what you will compare or predict, and who on the team does what. |
| 📊 **Data sources** | Your project data set (already filled in), plus any other data you will use. Say what each is and link it. |
| 🧹 **Data wrangling** | What the data needs before analysis: choosing columns, filtering with `.where`, turning dates or text into numbers, `group`, joining tables, missing values. |
| 🔬 **Planned tests** | For each question: the null hypothesis, the alternative, and the test statistic. For a regression or nearest-neighbor question: what you will predict, from what, and how you will judge it. |
| 🧰 **Toolbox** | Which course tools you will use (see below). |

### 📖 Finding and citing the literature
- **Where to search:** [Google Scholar](https://scholar.google.com) and [Temple University Libraries](https://library.temple.edu). Search for your variables and your topic together, e.g. *urban stream discharge rainfall* or *gut microbiome diversity inflammatory bowel disease*.
- **What counts:** peer-reviewed journal articles, review articles, and reports from scientific or government agencies (USGS, EPA, CDC, NOAA). A news story or a website summary can point you to a study, but cite the study itself.
- **Read the abstract first.** Keep the sources that say something about your question, your variables, or your method.
- **Cite it in one consistent style**, for example APA: *Paul, M. J., & Meyer, J. L. (2001). Streams in the urban landscape. Annual Review of Ecology and Systematics, 32, 333–365. https://doi.org/10.1146/annurev.ecolsys.32.081501.114040*

### 🧰 The data science toolbox
| Tool | Use it when | Read more |
|---|---|---|
| 🎲 **Hypothesis test with a simulated p-value** | You compare two groups, or a sample with a model: *is the difference real, or chance?* | [Inferential Thinking 11](https://inferentialthinking.com/chapters/11/Testing_Hypotheses.html), [12.1 A/B Testing](https://inferentialthinking.com/chapters/12/1/AB_Testing.html) |
| 📈 **Regression with bootstrap confidence intervals** | You predict one number from another, and want to know how sure you can be of the slope. | [Inferential Thinking 15](https://inferentialthinking.com/chapters/15/Prediction.html), [16](https://inferentialthinking.com/chapters/16/Inference_for_Regression.html), [13.2 The Bootstrap](https://inferentialthinking.com/chapters/13/2/Bootstrap.html) |
| 🧭 **Nearest neighbors** | You classify something into a category, or predict a value, from similar examples. | [Inferential Thinking 17, Classification](https://inferentialthinking.com/chapters/17/Classification.html) |

**Before you start writing, discuss as a team:** What do you want to find out? Which columns in your data set measure it? What would the answer look like if there were *no* effect?
