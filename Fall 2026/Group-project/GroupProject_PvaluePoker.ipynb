## How it works
Each round, your instructor **deals a hand**: a short story, a null hypothesis, and a histogram of what the test statistic looks like in thousands of simulations **if the null hypothesis is true**. A red line marks what was actually observed.

1. **Estimate the p-value** as a team: the chance, under the null hypothesis, of a result *at least as extreme* as the red line. That is the **area of the histogram at or beyond the line** (the bars are in percent, so you can add them up).
2. **One teammate places the bet** (a percent from 0 to 100). Your team can change it until the reveal.
3. **The reveal:** the closest bet earns 5 points, then 3, then 2; every other team that bet earns 1. 🎯 A bet within 1 percentage point earns a bonus point.
4. **Check it yourself:** the reveal shows the simulation code. Copy it into that round's cell below and run it.

You play for your project team from *GroupProject_Teams*; there is nothing to sign in to.
