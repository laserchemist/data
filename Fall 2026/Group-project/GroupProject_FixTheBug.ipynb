## 🐛 Bug 5 · `for` loop — count the high-expression samples
This cell should walk through every row of `genes` and count the samples with `Expression` above 8, saving the count as `high_count`.

```python
expression = genes.column('Expression')
high_count = 0
for i in genes.num_rows:
    if expression.item(i) > 8:
        high_count = high_count + 1
print('Samples with expression above 8:', high_count)
```

## 🐛 Bug 7 · Bootstrap — how much could the tumor average vary?
This cell should bootstrap the average tumor expression 500 times and save the 500 averages as `boot_means`. It runs with **no error**. Look at the printed interval: can that be right?

```python
boot_means = make_array()
for i in np.arange(500):
    resample = tumor.sample(with_replacement=False)
    boot_means = np.append(boot_means, np.mean(resample.column('Expression')))

left, right = percentile(2.5, boot_means), percentile(97.5, boot_means)
print('95% confidence interval for the average tumor expression:', left, 'to', right)
```
