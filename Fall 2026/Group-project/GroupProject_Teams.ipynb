<details>
<summary><b>🎲 How does the wheel keep the teams equal?</b> (click to open)</summary>

The wheel is a random choice, but not from *all* the teams: only from the teams that have the **fewest members so far**. With the tools from class, the idea looks like this:

```python
sizes = Table().with_columns('Team',    make_array('Data Owls', 'JAWN', 'Plotting Phils'),
                             'Members', make_array(3, 2, 2))
smallest = sizes.where('Members', min(sizes.column('Members'))).column('Team')
np.random.choice(smallest)          # 'JAWN' or 'Plotting Phils', each with chance 1/2
```

So the *team* you land on is random, but the team *sizes* never differ by more than about one. Assigning people to groups at random, instead of letting them choose, is the same idea used to set up fair experiments: see Inferential Thinking, [2.4 Randomization](https://inferentialthinking.com/chapters/02/4/Randomization.html), and [Chapter 9, Randomness](https://inferentialthinking.com/chapters/09/Randomness.html) for `np.random.choice`.

**Think about it:** if 58 students spin onto 7 teams, what team sizes do you expect?
