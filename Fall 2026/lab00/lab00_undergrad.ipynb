## Part 3 — Variables

A **variable** is a labeled container that stores a value so you can reuse it later — exactly
like recording a measurement in a lab notebook instead of trusting your memory. You create one
with `=`:

```python
mass_g = 12.4          # stores 12.4 in a variable called "mass_g"
sample_id = "A-3"      # stores text (a "string") in a variable called "sample_id"
```

Descriptive variable names (`mass_g` instead of `x`) make your code — and your future self —
much easier to follow.

## Part 4 — Python as a Calculator

Python evaluates arithmetic exactly like a calculator, which makes it useful for anything from
unit conversions to propagating uncertainty:

| Symbol | Meaning | Example |
|--------|---------|---------|
| `+` | add | `3 + 4` → 7 |
| `-` | subtract | `10 - 4` → 6 |
| `*` | multiply | `3 * 7` → 21 |
| `/` | divide | `10 / 4` → 2.5 |
| `**` | exponent | `2 ** 4` → 16 |

````python
# Python can do math!
print(5 + 3)
print(10 - 4)
print(3 * 7)
print(2 ** 4)   # 2 to the power of 4
````
