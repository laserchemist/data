### 🧰 Regression Toolbox
| Function | What it returns |
|---|---|
| `standard_units(any_numbers)` | the array converted to standard units: (value − mean) / SD |
| `correlation(t, label_x, label_y)` | the correlation coefficient *r* between two columns of Table `t` |
| `slope(t, label_x, label_y)` | slope of the regression line, $r \cdot \frac{SD_y}{SD_x}$, in original units |
| `intercept(t, label_x, label_y)` | intercept of the regression line, $\bar{y} - \text{slope} \cdot \bar{x}$ |
| `rmse(actual, predicted)` | root mean squared error, in the same units as the target |

````python
def standard_units(any_numbers):
    "Convert an array of numbers to standard units."
    return (any_numbers - np.mean(any_numbers)) / np.std(any_numbers)

def correlation(t, label_x, label_y):
    "The correlation coefficient r between two columns of Table t."
    return np.mean(standard_units(t.column(label_x)) * standard_units(t.column(label_y)))

def slope(t, label_x, label_y):
    "Slope of the regression line, in original units."
    r = correlation(t, label_x, label_y)
    return r * np.std(t.column(label_y)) / np.std(t.column(label_x))

def intercept(t, label_x, label_y):
    "Intercept of the regression line, in original units."
    return np.mean(t.column(label_y)) - slope(t, label_x, label_y) * np.mean(t.column(label_x))

def rmse(actual, predicted):
    "Root mean squared error, same units as the target."
    return np.sqrt(np.mean((actual - predicted) ** 2))
````

#### Try it: boiling points of the straight-chain alkanes
Fit a line predicting boiling point (K) from the number of carbons, then check how far off the line is on average. Swap in your own Table and column labels to reuse the toolbox.

````python
alkanes = Table().with_columns(
    'name',     make_array('methane', 'ethane', 'propane', 'butane', 'pentane', 'hexane', 'heptane', 'octane'),
    'carbons',  make_array(1, 2, 3, 4, 5, 6, 7, 8),
    'bp',       make_array(111.7, 184.6, 231.1, 272.7, 309.2, 341.9, 371.6, 398.8))

r = correlation(alkanes, 'carbons', 'bp')
m = slope(alkanes, 'carbons', 'bp')
b = intercept(alkanes, 'carbons', 'bp')
fitted = m * alkanes.column('carbons') + b
print(f'r = {r:.3f}   bp = {m:.1f} * carbons + {b:.1f}   RMSE = {rmse(alkanes.column("bp"), fitted):.1f} K')

alkanes.scatter('carbons', 'bp')
plt.plot(alkanes.column('carbons'), fitted, color='red', lw=2, label='regression line')
plt.legend();
````

> 🔎 **Look before you trust *r*.** An *r* near 1 does not mean the relationship is a straight line. Plot the residuals (`alkanes.column('bp') - fitted`) against `carbons` — a curved pattern means a line is not the best model ([15.5 Visual Diagnostics](https://inferentialthinking.com/chapters/15/5/Visual_Diagnostics.html)).

### 🧰 kNN Toolbox
| Function | What it returns |
|---|---|
| `distance(pt1, pt2)` | Euclidean distance between two points given as arrays |
| `distances(training, example, features)` | `training` with a new `'Distance'` column (distance from each row to `example`) |
| `closest(training, example, k, features)` | the `k` rows of `training` nearest to `example` |
| `predict_knn(training, example, k, features, target)` | **regression:** average `target` of the `k` nearest neighbors |
| `predict_all(training, evaluate, k, features, target)` | array of kNN predictions, one for every row of `evaluate` |
| `rmse_knn(training, evaluate, k, features, target)` | RMSE of the kNN predictions for `evaluate` |
| `classify_knn(training, example, k, features)` | **classification:** most common `'class'` among the `k` nearest neighbors |
| `standardize(t, features, ruler)` | copy of `t` with each feature in standard units, using the mean and SD of `ruler` (the training set) |

**Workflow:** split → standardize → predict → score
1. `train, test = data.split(n)` — after `np.random.seed(...)` so results are reproducible
2. `train_su = standardize(train, features, train)` and `test_su = standardize(test, features, train)` — always measure with the **training** ruler
3. Choose `k` by comparing `rmse_knn(...)` (regression) or accuracy (classification) for several values of `k`

````python
def distance(pt1, pt2):
    "The Euclidean distance between two points, given as arrays."
    return np.sqrt(np.sum((pt1 - pt2) ** 2))

def distances(training, example, features):
    "Training Table with a 'Distance' column: distance from each training row to example."
    dists = make_array()
    for row in training.select(features).rows:
        dists = np.append(dists, distance(np.array(row), np.array(example)))
    return training.with_columns('Distance', dists)

def closest(training, example, k, features):
    "The k rows of training that are nearest to example."
    return distances(training, example, features).sort('Distance').take(np.arange(k))

def predict_knn(training, example, k, features, target):
    "Predict the target for example: the average target of its k nearest neighbors."
    return np.mean(closest(training, example, k, features).column(target))

def predict_all(training, evaluate, k, features, target):
    "An array of kNN predictions, one for every row of the Table evaluate."
    predictions = make_array()
    for example in evaluate.select(features).rows:
        predictions = np.append(predictions, predict_knn(training, example, k, features, target))
    return predictions

def rmse_knn(training, evaluate, k, features, target):
    "Root mean squared error of kNN predictions for every row of evaluate."
    predictions = predict_all(training, evaluate, k, features, target)
    errors = evaluate.column(target) - predictions
    return np.sqrt(np.mean(errors ** 2))

def classify_knn(training, example, k, features):
    "The most common class among the k nearest neighbors of example (training needs a 'class' column)."
    votes = closest(training, example, k, features).group('class').sort('count', descending=True)
    return votes.column('class').item(0)

def standardize(t, features, ruler):
    "Copy of Table t with each feature in standard units, using the mean and SD of `ruler` (the training set)."
    result = t
    for label in features:
        mean = np.mean(ruler.column(label))
        sd = np.std(ruler.column(label))
        result = result.with_columns(label, (t.column(label) - mean) / sd)
    return result
````

#### Try it: predict a boiling point from its neighbors
Use the `alkanes` Table from the regression example as the training set and predict nonane (9 carbons, actual bp 424.0 K). Which `k` works best here, and why does kNN struggle at the **edge** of the training data?

````python
features = make_array('carbons')
nonane = make_array(9)

closest(alkanes, nonane, 3, features).show()
for k in make_array(1, 2, 3):
    print(f'k = {k}: predicted bp = {predict_knn(alkanes, nonane, k, features, "bp"):.1f} K   (actual 424.0 K)')
````
