## 1. Assign projects — `assign()`
**Before:** upload one roster per class into **this folder on the hub**, named `class_<section>.csv` (e.g. `class_03.csv`). Both export layouts work: the registrar export with no header, and the Canvas export with a `Student, SIS Login ID, Section` header. 🔒 Rosters never go on GitHub.

`assign()` is **safe to rerun**:
- Students keep their saved project; new students on a roster join the smallest groups; students removed from a roster are removed.
- A class with no roster file here is **kept exactly as saved**.
- Each save first backs up the previous groups.

**Options**
- `assign(redeal='03')` deals class 03 again from scratch (only before students have seen their groups).
- `assign(drop='99')` deletes a class, e.g. a test class.

## 8. Discussion — `compare_lines(...)`
Students on the same project used the same data, so their full-data *r* and slope **should agree**. A wide range usually means a toolbox bug, a subset of the data (the twist line), or swapped *x* and *y* ([Ch. 15.2](https://inferentialthinking.com/chapters/15/2/Regression_Line.html)).

## 9. Canvas grades — `export_canvas(...)`
Writes a gradebook file for Canvas: **5 points for every complete submission**, 0 for everyone else. *Complete* means a figure linked to the student, with a caption, that passes the figure check (axis labels, title, scatter points, regression line, legend). Any project counts; reflections aren't graded. Students who aren't complete are listed with what's missing, so you can check before importing.

`export_canvas('03')` writes `canvas_microproject_03.csv` for one class; `'all'` writes one file for every class. Then in Canvas: **Grades → Import**, choose the file, and confirm the new assignment column.

**Options**
- `incomplete=''` leaves incomplete students' Canvas grade unchanged instead of 0.
- `points=10` changes the score.
- `strict_figure=False` only requires a figure and a caption.
- `require_reflection=True` also requires a reflection.
- `assignment='Regression Microproject (123456)'` updates an assignment you already made in Canvas: copy its exact column header from a Canvas gradebook export.

🔒 The file holds names and logins: download it, import it, and keep it off GitHub.

```python
export_canvas('all')
```
