```python
lunch = make_array('pizza','steak','tuna','yogurt','Chik-Fil-a')
```

### Compute your lunch cost with conditional

```python
if eat == 'pizza':
    cost = 4
elif eat == 'steak':
    cost = 11
elif eat == 'tuna':
    cost = 7
elif eat == 'yogurt':
    cost = 4.5
elif eat == 'Chik-Fil-a':
    cost = 15
else:
    cost = 0
```

```python
print(f'You cost for eating {eat} for lunch today is ${cost}!')
```

### Improved scoring

```python
realistic_scores = [0, 3, 6, 7, 10, 13, 14, 17, 20, 21,
                    23, 24, 27, 28, 30, 31, 34, 35, 38, 41, 42, 45]
```

#### Choose from realistic scores

```python
es = np.random.choice(realistic_scores)
cs = np.random.choice(realistic_scores)
```

```python
print(f'Sample game, Eagles {es}: Cowboys {cs}')
```

```python
if es > cs:
    print('WIN')
elif es == cs:
    print('TIE')
else:
    print('Lose')
```

```python
simscore = Table().with_columns('Eagles', np.random.choice(realistic_scores,5000),
                                'Cowboys', np.random.choice(realistic_scores,5000))
simscore
```

```python
simscore=simscore.with_columns('Eagles Win', simscore.column('Eagles')>simscore.column('Cowboys'))
simscore
```

```python
np.sum(simscore.column('Eagles Win'))/simscore.num_rows*100
```

```python
simscore.column('Eagles')>simscore.column('Cowboys')
```
