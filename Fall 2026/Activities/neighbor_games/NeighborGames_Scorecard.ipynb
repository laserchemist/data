### 🚦 Start the competition — `stoplight('neighbor_games')`
The Neighbor Games start on **🔴 red**: students can run their notebooks, but the **Submit buttons don't send anything**. A student who clicks Submit early sees *The competition hasn't started yet… click Submit again after GO*, so nobody gets a head start on the board while other teams are still getting set up.
- **🟡 Get ready** — submissions are still closed.
- **🟢 GO! (3-2-1)** — counts down, opens submissions, and starts the clock shown in the leaderboard header.
- **🔴 Stop** after GO — *time's up*: no more submissions; the board keeps everyone's last scores. Click GO again to reopen.

The light is saved in the shared folder, so it's the same for every team. `reset_game('neighbor_games')` turns it back to red for the next class.

```python
stoplight('neighbor_games')
```

## 📂 Earlier classes — `list_archives()` and `restore_game(...)`
`reset_game(...)` never deletes a class: it moves its board to an archive. **`list_archives('neighbor_games')`** shows every class you can bring back, and how many files the board holds right now.

**`restore_game('neighbor_games', '01')`** puts class 01's board back, from its newest archive, and sets the class to 01. Then use the cells above as usual: `neighbor_scoreboard()` to show it, `save_overview('neighbor_games')` for `neighbor_games_01.html`, `neighbor_canvas('01')` for Canvas.
- Whatever is on the board now is **archived first** (as the class that's set), so nothing is lost. The archive you restore from stays as it is.
- With no archive for that class, it rebuilds the board from the class's saved `neighbor_games_01_…_scores.csv` and `_teams.csv` in this folder. Canvas then matches students by the team-member names only.
- `restore_game('neighbor_games', '01', archive='neighbor_games_01_20261104-1032')` picks an older archive from the list.

When you're done, `reset_game('neighbor_games')` clears the board again (it saves and re-archives class 01), then `set_section(...)` for the next class.

## 💾 End of class — save, then reset
**`save_results('neighbor_games')`** writes this class's results to CSV files in this folder, named with the section and date, e.g. `neighbor_games_08_2026-11-04_scores.csv`. It writes `…_scores.csv` and `…_teams.csv` (each team's members and the rounds it submitted), **`neighbor_games_08.html`**, the final board as a web page to show later (right-click → Open in New Browser Tab, or download it), and the Canvas file when `class_08.csv` is in this folder. Download them from the file browser.

**`reset_game('neighbor_games')`** clears the boards for the next class. It **saves the CSV files first, automatically**, so you can't lose a class's results. It then archives the raw files (nothing is deleted) and turns the start light back to red. Then set the next class with `set_section(...)`. Remove the `#` to reset.

🔒 The CSV files hold students' names: keep them off GitHub.
