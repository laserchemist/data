### 🚦 Start the competition — `stoplight()`
The Smackdown starts on **🔴 red**: students can spin and load their data, but **check-ins are closed**. Their check-in cells say the competition hasn't started and to run the cell again after GO, so nobody gets a head start on the board while others are still finding their teams.
- **🟡 Get ready** — teams are together; check-ins are still closed.
- **🟢 GO! (3-2-1)** — counts down, opens check-ins, and starts the clock. The ⚡ finish times on the standings are measured from GO.
- **🔴 Stop** after GO — *time's up*: check-ins close, and the boards show the final time. Students can still post their Step 8 summaries. Click GO again to reopen.

The light is saved in the shared folder, so it's the same for every student, and all the boards show it in their headers. `reset_game(...)` turns it back to red for the next class.

```python
stoplight()
```

Each row shows how many teammates have checked in, who has finished, and the team's slope and 95% CI, ready for the discussion questions.

### 🤝 Team race — `smackdown_team_race()`
**The team wins when everyone on it finishes.** This board shows each team's progress toward a *collective* finish, with **no student names**, so it's safe to keep on the projector:
- **One block per teammate**, filled by that person's points and turned green with ✓ at all 13. Teammates who spun onto the team but haven't checked in yet show as empty blocks: they count.
- **The step strip** shows how many teammates have each step (e.g. *Bootstrap 2/4*), so a team can see where to help.
- **% of team points** is everyone's points over 13 × team size. Teams with every member finished go to the top with 🏁 and their time after GO; the first one gets the 🏆 banner.

**Projector view:** right-click `bootstrap_smackdown_team_race.html` → **Open in New Browser Tab**.

```python
smackdown_team_race()
```
